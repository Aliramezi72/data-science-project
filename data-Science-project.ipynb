```python
import matplotlib as plt
from matplotlib import pyplot
import numpy as np
```

```python
bins = np.linspace(min(df["hp_kW"]), max(df["hp_kW"]), 4)
bins
```

```python
group_names = ['Low', 'Medium', 'High']
```

```python
df['hp_kW_binned']= pd.cut(df['hp_kW'], bins, labels=group_names, include_lowest=True )
df[['hp_kW','hp_kW_binned']].head(20)
```

```python
df['hp_kW_binned'].value_counts()
```

```python
# Data visualization, but not efficient.
plt.pyplot.hist(df["hp_kW"])
plt.pyplot.xlabel("hp_kW")
plt.pyplot.ylabel("count")
plt.pyplot.title("hp_kW")
```

```python
pyplot.bar(group_names, df["hp_kW_binned"].value_counts())
plt.pyplot.xlabel("hp_kW")
plt.pyplot.ylabel("count")
plt.pyplot.title("hp_kW_binned")
```

```python
import matplotlib.pyplot as plt

counts = df["hp_kW_binned"].value_counts().sort_index()

plt.scatter(x=counts.index, y=counts.values)

plt.xlabel("hp_kW")
plt.ylabel("count")
plt.title("hp_kW_binned")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np


counts = df["hp_kW_binned"].value_counts().reindex(group_names)


plt.bar(group_names, counts, color='skyblue', edgecolor='black')


plt.plot(group_names, counts, color='red', marker='o', linestyle='-', linewidth=2, label='Trend Line')


plt.xlabel("hp_kW Category")
plt.ylabel("Count")
plt.title("hp_kW Binned Histogram with Trend Line")
plt.legend()

plt.show()
```
